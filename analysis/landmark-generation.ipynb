```python
import random

LN = 3
zone_start = 10
zone_end = zone_start + 10*LN
zone_length = zone_end - zone_start
maxSpacing = zone_length / (LN+1)

def one_landmark_gen(variation):
        spacing = maxSpacing * (1.0 + random.uniform(-variation, variation))


        total_width =(LN-1) *spacing
        max_start = zone_end - total_width

        start = random.uniform(zone_start, max_start)
        pos =[]
        for i in range(LN):
            pos.append(start +(i)*spacing)
        return pos, spacing

# define the constants
def gen_trials(var):
    num_trials = 1000
    trials = []
    spacing = []
    for i in range(0, num_trials):
        positions, s = one_landmark_gen(var)
        trials.append(positions)
        spacing.append(s)
    
    return trials, spacing
```

variation changes how much the spacing changes

```python
variation = 0.8

trials, spacing = gen_trials(variation)

import matplotlib.pyplot as plt

# Separate each landmark column
L1 = [row[0] for row in trials]
L2 = [row[1] for row in trials]
L3 = [row[2] for row in trials]

plt.figure()

# Overlay histograms (true distribution view)
plt.hist(L1, bins=20, alpha=0.5, label="Landmark 1")
plt.hist(L2, bins=20, alpha=0.5, label="Landmark 2")
plt.hist(L3, bins=20, alpha=0.5, label="Landmark 3")

plt.xlabel("Position")
plt.ylabel("Frequency")
plt.title("Distribution of Landmark Positions")
plt.legend()

plt.show()
```
