```python
import math
import numpy as np
import matplotlib.pyplot as plt



def sensor(distance, f, g):
    return 1 / (1 + math.exp(f * (g * distance - 1)))

# generate distance values
distances = np.arange(0, 3.1, 0.01)
threshold = 0.00001

f_val = [8]
g_val = [1]
# make figure bigger
plt.figure(figsize=(10, 6))

for x in g_val:
    for i in f_val:  # start from 1 (f=0 gives constant 0.5)
        # compute sensor outputs
        outputs = [sensor(d, i, x) for d in distances]

        # plot curve and capture its colour
        line, = plt.plot(distances, outputs, label=f"f = {i}, h = {x}")
        colour = line.get_color()

        # # find cutoff distance
        # for d in distances:
        #     if sensor(d, i, x) < threshold:
        #         cutoff = d
        #         plt.axvline(x=d, linestyle="--", color=colour)
        #         # add text next to the line
        #         plt.text(
        #             cutoff + 0.03,   # small shift right so text isn't on top of line
        #             0.5,             # vertical position (middle of plot)
        #             f"cutoff ≈ {cutoff:.2f}",
        #             rotation=90,
        #             verticalalignment="center",
        #             color=colour
        #         )
                # break

plt.xlabel("Distance")
plt.ylabel("Sensor Output")
plt.title(f"Sensor Output f = {8}, h = {1}")
plt.legend()
plt.grid(True)
plt.show()
```
